## Key terms to know first


<div style="padding: 5px; padding-left: 10px; background:#e6e6e6">
A few words you will hear throughout the session.
<p><b>Large language model, or LLM.</b> A program trained on large amounts of text that answers by predicting likely next pieces of text. ChatGPT, Claude, and Gemini are all built on LLMs.</p>
<p><b>Token.</b> A small chunk of text, a word or part of a word, that the model reads and writes one piece at a time.</p>
<p><b>Context window.</b> How much text the model can consider at once. In a very long conversation, earlier details can fall out of view and the model may lose track of them.</p>
<p><b>Prompt.</b> The message you send the model. A clearer prompt gives a clearer answer.</p>
<p><b>Hallucination.</b> A confident, fluent answer that is actually wrong, such as an invented function or a made-up citation. This is why you always verify.</p>
</div>

## 2. Writing a good prompt


AI gives a better answer when you give it a better question. You do not need special tricks. You need to say who you are, what you are looking at, and exactly what you want.

<div style="border-left: 3px solid #000; padding: 1px; padding-left: 10px; background: #F0FAFF; ">
<b>The parts of a good prompt</b>
<ul>
<li>State your level, for example that you are new to Python.</li>
<li>Give context, meaning what the code or data is and what you are trying to do.</li>
<li>Make a specific ask, such as explain line by line, find the bug, or show one alternative.</li>
<li>Say how you want help, for example help me understand rather than just give the answer, or keep it short.</li>
<li>Ask it to check itself. If it is unsure, it should say so and point you to the official documentation.</li>
</ul>
</div>

## 3. When to lean on AI, and when to work it out yourself


AI makes it easy to never be stuck, and that is the danger. A skill you hand off before you have built it does not develop. The real skill is not using AI or avoiding it. It is noticing which situation you are in.

<div style="border-left: 3px solid #000; padding: 1px; padding-left: 10px; background: #F0FAFF; ">
<b>A rough decision guide</b>
<table>
<tr><th align="left">Work it out first, to build the skill</th><th align="left">Hand it to AI, to save the time</th></tr>
<tr><td>It is a core idea you are here to learn, such as indexing, loops, or logic</td><td>Boilerplate you already understand but find tedious, such as plot styling or file paths</td></tr>
<tr><td>You have been at it about 5 to 15 minutes and are making progress</td><td>You have been genuinely stuck for more than 15 to 20 minutes with no progress</td></tr>

## 5. Plot spike waveforms, then use AI to make the figure publication quality


You will make a real figure and use AI to bring it up to publication quality, one change at a time. This is how AI fits into everyday research. You make a first plot, then refine it, and you check each step.

<div style="border-left: 3px solid #000; padding: 1px; padding-left: 10px; background: #F0FAFF; ">
<b>About the extracellular action potential</b>
<p>An electrode just outside a neuron records a spike whose main deflection points down, unlike the upward intracellular spike.</p>
<p>Spike width is the time from the negative trough to the positive peak that follows. Narrow spikes are often fast-spiking inhibitory interneurons, and broad spikes are often pyramidal cells. The two waveforms below share the same trough depth, so the width is what stands out.</p>
</div>

```python
import numpy as np
import matplotlib.pyplot as plt

# Time axis in milliseconds
t = np.linspace(-1.0, 2.0, 601)

# Two synthetic extracellular spikes, both pointing down, normalized to the same trough depth.
# Narrow spike, about 300 microseconds trough to peak, often a fast-spiking interneuron.
interneuron = (10 * np.exp(-((t + 0.12) / 0.05) ** 2)
               - 100 * np.exp(-((t) / 0.05) ** 2)
               + 60 * np.exp(-((t - 0.30) / 0.13) ** 2))
interneuron = interneuron / abs(interneuron.min()) * 100

# Broad spike, about 700 microseconds trough to peak, often a pyramidal cell.
pyramidal = (14 * np.exp(-((t + 0.26) / 0.11) ** 2)
             - 100 * np.exp(-((t) / 0.12) ** 2)
             + 45 * np.exp(-((t - 0.70) / 0.34) ** 2))
pyramidal = pyramidal / abs(pyramidal.min()) * 100

# A plain starting figure. You will use AI to bring it up to publication quality.
plt.plot(t, interneuron)
plt.plot(t, pyramidal)
plt.xlabel("Time (ms)")
plt.ylabel("Voltage (uV)")
plt.show()
```

Run the cell below to measure each trough-to-peak width from the data, in microseconds. Check that the widths in your legend match, and that each marker sits on the true trough and peak.

```python
def trough_peak(w):
    trough_idx = np.argmin(w)
    peak_idx = trough_idx + np.argmax(w[trough_idx:])
    width_us = (t[peak_idx] - t[trough_idx]) * 1000   # milliseconds to microseconds
    return trough_idx, peak_idx, width_us

for name, w in [("narrow-spiking (interneuron)", interneuron),
                ("broad-spiking (pyramidal)", pyramidal)]:
    ti, pi, width_us = trough_peak(w)
    print(f"{name}: trough {t[ti]:.3f} ms, peak {t[pi]:.3f} ms, width {width_us:.0f} us")
# Expected  narrow about 300 us, broad about 700 us
```

<div style="border-left: 3px solid #000; padding: 1px; padding-left: 10px; background: #F0FAFF; ">
<b>Why the width matters</b>
<p>Trough-to-peak width is one of the main features used to separate cell types in extracellular recordings. Narrow spikes tend to be putative fast-spiking interneurons, and broad spikes tend to be putative pyramidal cells. You will see trough-to-peak width again in the Cell Types lesson.</p>

## 6. Guardrails that keep your work rigorous


AI-assisted coding in science has real risks. It can invent functions, write tests that check nothing, and produce code that runs but is wrong for your method. Bridgeford and colleagues (2026) offer twelve quick tips for keeping AI-assisted coding rigorous. Here is the full set as a reference to keep.

<div style="border-left: 3px solid #000; padding: 1px; padding-left: 10px; background: #F0FAFF; ">
<b>Twelve quick tips for AI-assisted coding in science</b> (Bridgeford et al., 2026)
<table>
<tr><th align="left">Stage</th><th align="left">Tip</th></tr>
<tr><td rowspan="3"><b>Prepare and understand</b></td><td>1. Gather background knowledge before implementation</td></tr>
<tr><td>2. Define your problem, and separate problem framing from coding</td></tr>
<tr><td>3. Choose appropriate AI interaction models</td></tr>
<tr><td rowspan="2"><b>Plan and give context</b></td><td>4. Think through and divide an implementation plan</td></tr>
<tr><td>5. Manage context strategically</td></tr>
<tr><td rowspan="3"><b>Test and validate</b></td><td>6. Meticulously verify AI assistance for testing</td></tr>
<tr><td>7. Use tests to specify and constrain AI-generated code</td></tr>
<tr><td>8. Leverage AI to audit and expand test coverage</td></tr>
<tr><td rowspan="4"><b>Review and finalize</b></td><td>9. Monitor progress, and know when to restart</td></tr>
<tr><td>10. Critically review generated code</td></tr>
<tr><td>11. Refine code incrementally with focused objectives</td></tr>
<tr><td>12. Document your code for reuse and publication</td></tr>
</table>
<p>One idea sits under all twelve. "AI wrote it" is never a valid defense for a flawed result. You remain accountable for your code.</p>
</div>

<div style="border-left: 3px solid #000; padding: 1px; padding-left: 10px; background: #F0FAFF; ">
<b>What not to paste into a public AI tool</b>
<p>Free and consumer AI tools may store or reuse what you type. Before you paste, think about what the text contains. Avoid pasting the following.</p>
<ul>
<li>Unpublished or embargoed data and results</li>
<li>Human-subjects data, anything with personal or health identifiers, or data covered by IRB or a data use agreement</li>
<li>Proprietary, licensed, or confidential code</li>
</ul>
<p>Safer habits help. Paste a small made-up example that reproduces the problem, use de-identified or synthetic data, and check your institution's policy and any data use agreements. When in doubt, share the shape of the problem rather than the real data.</p>
</div>
