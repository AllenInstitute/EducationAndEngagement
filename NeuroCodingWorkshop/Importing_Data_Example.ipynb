# Does brain size scale with the number of neurons?

## Step 1. Import our data

```python
import pandas as pd

df = pd.read_csv('species_brainmass_neurons.csv')

df.head()
```

## Step 2. Subset the data into primates and non-primates

```python
primate_df = df[df['Order']=='Primata']
nonprimate_df = df[df['Order']!='Primata']

primate_df.head()
```

```python
import matplotlib.pyplot as plt

fig , ax = plt.subplots(1,2,figsize=(10,5))

x_1 = primate_df['cortex_mass_g']
y_1 = primate_df['Neurons']

x_2 = nonprimate_df['cortex_mass_g']
y_2 = nonprimate_df['Neurons']

ax[0].scatter(x_1,y_1)
ax[1].scatter(x_2,y_2)

plt.xscale('log')
plt.yscale('log')
```
